```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv('/Users/janoskonig/Downloads/MAI-chewingtoday.csv')
```

```python
# Convert chewing_today_situation to numeric

# data['chewing_today_situation'] = data['chewing_today_situation'].replace({
#     'Nagyon rossz': 1,
#     'Rossz': 2,
#     'Átlagos': 3,
#     'Jó': 4,
#     'Kiváló': 5
# })
```

```python
# Drop the init_mai NaN values
data = data.dropna(subset=['init_mai'])
```

```python
data
# Export to CSV
```

```python
# Compute the correlation value (Pearson)
pearson_correlation = data['chewing_today_situation'].corr(data['init_mai'], method='pearson')
print(f"Correlation between chewing_today_situation and init_mai: {pearson_correlation:.2f}")
# Interpret the correlation value
if pearson_correlation > 0.5:
    print("Strong positive correlation")
elif pearson_correlation > 0:
    print("Weak positive correlation")
elif pearson_correlation < -0.5:
    print("Strong negative correlation")
elif pearson_correlation < 0:
    print("Weak negative correlation")


# Compute the correlation value (Spearman)
spearman_correlation = data['chewing_today_situation'].corr(data['init_mai'], method='spearman')
print(f"Spearman correlation between chewing_today_situation and init_mai: {spearman_correlation:.2f}")
# Interpret the Spearman correlation
if spearman_correlation > 0.5:
    print("Strong positive correlation")
elif spearman_correlation > 0:
    print("Weak positive correlation")
elif spearman_correlation < -0.5:
    print("Strong negative correlation")
elif spearman_correlation < 0:
    print("Weak negative correlation")
    
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set the style for the plot
sns.set(style="whitegrid")

# Create a scatter plot with a regression line
plt.figure(figsize=(8, 6))
plt.scatter(data['chewing_today_situation'], data['init_mai'])
plt.title('Korreláció a páciens szubjektív véleménye és a MAI között')
plt.xlabel('Szubjektív rágóképesség (minél magasabb, annál jobb)')
plt.ylabel('Mixing Ability Index (minél alacsonyabb, annál jobb)')
# Write the correlation value on the plot
plt.text(0.05, 0.95, f'Pearson r = {pearson_correlation:.2f}', ha='left', va='top', transform=plt.gca().transAxes)
plt.text(0.05, 0.90, f'Spearman r = {spearman_correlation:.2f}', ha='left', va='top', transform=plt.gca().transAxes)
plt.show()

# violin plot
plt.figure(figsize=(10, 4))
sns.violinplot(x='chewing_today_situation', y='init_mai', data=data, order=['Nagyon rossz', 'Rossz', 'Átlagos', 'Jó', 'Kiváló'], color='green')
plt.title('Az első mixing ability test és a szubjektív rágóképesség közötti korreláció')
plt.xlabel('Szubjektív rágóképesség')
plt.ylabel('Iniciális MAI')
plt.xticks(rotation=45)
plt.show()
```
