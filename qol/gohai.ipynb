```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
gohai = pd.read_csv('/Users/janoskonig/Downloads/gohai.csv')
```

```python
# Mean of GOHAI_1 to GOHAI_12
gohai['GOHAI'] = gohai[['GOHAI_1', 'GOHAI_2', 'GOHAI_3', 'GOHAI_4', 'GOHAI_5', 'GOHAI_6', 'GOHAI_7', 'GOHAI_8', 'GOHAI_9', 'GOHAI_10', 'GOHAI_11', 'GOHAI_12']].mean(axis=1)
```

```python
gohai
```

```python
# Violin plot of the responsiveness_today_situation and mean OHIP scores
plt.figure(figsize=(10, 6))
sns.violinplot(x='responsiveness_today_situation', y='GOHAI', data=gohai, inner='quartile', order=['Nagyon rossz', 'Rossz', 'Átlagos', 'Jó', 'Kiváló'])
plt.title('A páciens GOHAI átlagos pontszáma a jelen helyzet szubjektív megítélése viszonyában')
plt.xlabel('A jelen helyzet szubjektív megítélése')
plt.ylabel('GOHAI pontszám')
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
# Show the plot

```
