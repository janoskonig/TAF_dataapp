```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
ohip = pd.read_csv('/Users/janoskonig/Downloads/OHIP.csv')
```

```python
# Average the OHIP scores per participant
ohip['mean'] = ohip.iloc[:, 1:6].mean(axis=1)
```

```python
ohip
```

```python
# Convert the responsiveness_today_situation values to numbers
# Kiváló = 5, Jó = 4, Átlagos = 3, Rossz = 2, Nagyon rossz = 1
def convert_responsiveness(value):
    if value == 'Kiváló':
        return 5
    elif value == 'Jó':
        return 4
    elif value == 'Átlagos':
        return 3
    elif value == 'Rossz':
        return 2
    elif value == 'Nagyon rossz':
        return 1
    else:
        return np.nan
    
# ohip['responsiveness_today_situation'] = ohip['responsiveness_today_situation'].apply(convert_responsiveness)
```

```python
# Boxplot of the responsiveness_today_situation and mean OHIP scores
plt.figure(figsize=(10, 6))
sns.boxplot(x='responsiveness_today_situation', y='mean', data=ohip)
plt.title('Boxplot of Responsiveness Today Situation vs Mean OHIP Scores')
plt.xlabel('Responsiveness Today Situation')
plt.ylabel('Mean OHIP Score')
plt.xticks(rotation=45)
plt.grid(axis='y')
plt.tight_layout()
plt.show()

# Violin plot of the responsiveness_today_situation and mean OHIP scores
plt.figure(figsize=(10, 6))
sns.violinplot(x='responsiveness_today_situation', y='mean', data=ohip, order=['Nagyon rossz', 'Rossz', 'Átlagos', 'Jó', 'Kiváló'])
plt.title('A páciens OHIP-5 átlagos pontszáma a jelen helyzet szubjektív megítélése viszonyában')
plt.xlabel('Jelenleg megélt szájüregi egészséggel kapcsolatos életminőség')
plt.ylabel('Átlagos OHIP-5 pontszám')
plt.xticks(rotation=45)
plt.grid(axis='y')
plt.tight_layout()
plt.show()

# Beeswarm Plot
plt.figure(figsize=(10, 6))
sns.swarmplot(x='responsiveness_today_situation', y='mean', data=ohip, color='black', alpha=0.5)
plt.title('Beeswarm Plot of Responsiveness Today Situation vs Mean OHIP Scores')
plt.xlabel('Responsiveness Today Situation')
plt.ylabel('Mean OHIP Score')
plt.xticks(rotation=45)
plt.grid(axis='y')
plt.tight_layout()
plt.show()


```
